```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv("C:/Users/LENOVO/Downloads/iris_dataset.csv")

plt.figure(figsize=(8, 5))
plt.hist(
    df['sepal_length_cm'],
    bins=10,
    edgecolor='black',
    color='#2b5c8f',
    alpha=0.8,
)
plt.title('Histograma de Longitud del Sépalo (sepal_length_cm)')
plt.xlabel('Longitud del Sépalo (cm)')
plt.ylabel('Frecuencia')
plt.show()

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].hist(
    df['petal_length_cm'],
    bins=5,
    edgecolor='black',
    color='#d95f02',
    alpha=0.8,
)
axes[0].set_title('petal_length_cm (5 Bins)')
axes[0].set_xlabel('Longitud del Pétalo (cm)')
axes[0].set_ylabel('Frecuencia')

axes[1].hist(
    df['petal_length_cm'],
    bins=20,
    edgecolor='black',
    color='#7570b3',
    alpha=0.8,
)
axes[1].set_title('petal_length_cm (20 Bins)')
axes[1].set_xlabel('Longitud del Pétalo (cm)')
axes[1].set_ylabel('Frecuencia')
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df,
    x='species',
    y='sepal_length_cm',
    hue='species',
    palette='Set2',
    legend=False,
)
plt.title('Boxplot de Longitud del Sépalo por Especie')
plt.xlabel('Especie')
plt.ylabel('Longitud del Sépalo (cm)')
plt.show()

plt.figure(figsize=(9, 5))
num_cols = [
    'sepal_length_cm',
    'sepal_width_cm',
    'petal_length_cm',
    'petal_width_cm',
]
sns.boxplot(data=df[num_cols], palette='Pastel1')
plt.xticks(
    ticks=range(4),
    labels=['Sepal Len', 'Sepal Wid', 'Petal Len', 'Petal Wid'],
)
plt.title('Comparación de Variables Numéricas')
plt.ylabel('Medida (cm)')
plt.show()

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].scatter(
    df['sepal_length_cm'],
    df['sepal_width_cm'],
    color='#1b9e77',
    alpha=0.7,
    edgecolors='k',
)
axes[0].set_title('Scatter Plot: Sepal Length vs. Sepal Width')
axes[0].set_xlabel('Sepal Length (cm)')
axes[0].set_ylabel('Sepal Width (cm)')

axes[1].hist(
    df['sepal_length_cm'],
    bins=10,
    edgecolor='black',
    color='#e7298a',
    alpha=0.7,
)
axes[1].set_title('Histograma: Sepal Length')
axes[1].set_xlabel('Sepal Length (cm)')
axes[1].set_ylabel('Frecuencia')
plt.tight_layout()
plt.show()
```

1. Histograma: Longitud del sépalo (sepal_length_cm)
[file-tag: ej_1_histograma_sepal_length]

Descripción e Interpretación:

La mayor parte de las observaciones se concentra en el rango de 5.38 cm a 6.46 cm, registrando sus picos más altos en las barras de 5.38 – 5.74 cm (27 datos) y 6.10 – 6.46 cm (26 datos).

También se observa un pico secundario importante entre 4.66 y 5.02 cm (23 datos).

La distribución muestra una ligera asimetría hacia la derecha con valores que se extienden hasta los 7.9 cm, y una media general de 5.84 cm (mediana de 5.80 cm).

2. Histograma y Bins: petal_length_cm (5 vs. 20 Bins)
[file-tag: ej_2_histogramas_petal_length]

Cambio en la interpretación según los bins:

Con 5 bins: Brinda una visión general macro. Muestra claramente que la distribución es bimodal y que existe un vacío significativo entre los 2 cm y los 3 cm (que separa a la especie setosa de las demás).

Con 20 bins: Ofrece un nivel de detalle mucho mayor. Permite observar la estructura interna del segundo grupo (pétalos > 3 cm), revelando la presencia de pequeñas sub-poblaciones o variaciones en la densidad de datos entre los 4.0 cm y los 6.0 cm (correspondientes a versicolor y virginica).

3. Boxplot por Especie (sepal_length_cm)
[file-tag: ej_3_boxplot_por_especie]

Comparación visual de las medianas:

setosa: Presenta la mediana más baja, ubicada en 5.0 cm.

versicolor: Muestra una mediana intermedia de 5.9 cm.

virginica: Registrar la mediana más alta, situada en 6.5 cm.

Conclusión: Existe una progresión clara y ascendente en la longitud del sépalo desde setosa hasta virginica. Además, se observa un valor atípico (outlier) inferior en la especie virginica (alrededor de 4.9 cm).

4. Boxplot de Variables Numéricas
[file-tag: ej_4_boxplot_variables_numericas]

Problema de escala observado:

Aunque todas las variables comparten la misma unidad de medida (centímetros), existe una disparidad amplia en los rangos y la dispersión:

sepal_length_cm y petal_length_cm abarcan valores más altos (de 1.0 a 7.9 cm) y presentan mayor variabilidad (rango intercuartílico más ancho).

petal_width_cm y sepal_width_cm operan en rangos mucho más reducidos (de 0.1 a 4.4 cm).

En algoritmos de aprendizaje automático o análisis multivariado sin escalamiento previo (como estandarización o normalización), las variables con mayor magnitud (sepal_length_cm y petal_length_cm) dominarían sobre las de menor escala (petal_width_cm).
