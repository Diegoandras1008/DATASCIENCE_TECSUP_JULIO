```python
def procesar_ventas(lineas: list[str]) -> None:
    # Diccionarios nativos para acumulaciones
    conteo_estados = {}
    conteo_metodos_pago = {}
    unidades_por_producto = {}
    ingresos_por_producto = {}

    ingreso_total_entregado = 0.0
    ventas_arequipa = 0

    print("--- 4. TOTAL VENDIDO POR REGISTRO ---")

    for linea in lineas:
        linea_limpia = linea.strip()

        # Ignorar líneas vacías
        if not linea_limpia:
            continue

        # Separar por comas
        campos = [campo.strip() for campo in linea_limpia.split(",")]

        # VALIDACIÓN CLAVE: Si la línea no tiene 10 campos (por ejemplo la cabecera o un título), la ignoramos
        if len(campos) != 10 or campos[0] == "ID":
            continue

        (
            id_v,
            producto,
            talla,
            color,
            cliente,
            ciudad,
            cantidad_str,
            precio_str,
            metodo_pago,
            estado,
        ) = campos

        cantidad = int(cantidad_str)
        precio_unitario = float(precio_str)

        total_registro = cantidad * precio_unitario
        print(f"[{id_v}] {producto} -> Total: S/ {total_registro:.2f}")

        # Conteo de estados
        if estado in conteo_estados:
            conteo_estados[estado] += 1
        else:
            conteo_estados[estado] = 1

        # Conteo de métodos de pago
        if metodo_pago in conteo_metodos_pago:
            conteo_metodos_pago[metodo_pago] += 1
        else:
            conteo_metodos_pago[metodo_pago] = 1

        # Acumulación por producto
        if producto in unidades_por_producto:
            unidades_por_producto[producto] += cantidad
            ingresos_por_producto[producto] += total_registro
        else:
            unidades_por_producto[producto] = cantidad
            ingresos_por_producto[producto] = total_registro

        if estado == "entregado":
            ingreso_total_entregado += total_registro

        if ciudad.lower() == "arequipa":
            ventas_arequipa += 1

    print("--- 5. ESTADO DE LAS VENTAS ---")
    for est, total in conteo_estados.items():
        print(f"• {est.capitalize()}: {total}")

    print(
        f"\n--- 6. INGRESO TOTAL (ENTREGADOS) ---\nS/ {ingreso_total_entregado:.2f}"
    )

    # Buscar el producto más vendido
    prod_mas_vendido = None
    max_unidades = 0
    for prod, unids in unidades_por_producto.items():
        if unids > max_unidades:
            max_unidades = unids
            prod_mas_vendido = prod

    print(
        f"\n--- 7. PRODUCTO MÁS VENDIDO ---\n{prod_mas_vendido} ({max_unidades} unidades)"
    )

    print(f"\n--- 8. VENTAS EN AREQUIPA ---\nTotal: {ventas_arequipa} ventas")

    # Buscar el método de pago más usado
    metodo_top = None
    max_uso = 0
    for metodo, uso in conteo_metodos_pago.items():
        if uso > max_uso:
            max_uso = uso
            metodo_top = metodo

    print(
        f"\n--- 9. MÉTODO DE PAGO MÁS USADO ---\n{metodo_top} ({max_uso} transacciones)"
    )

    print("\n--- 10. RETO EXTRA: RESUMEN POR PRODUCTO ---")
    print(f"{'Producto':<25} | {'Unidades':<10} | {'Ingreso Generado':<15}")
    print("-" * 56)
    for prod, unids in unidades_por_producto.items():
        ing = ingresos_por_producto[prod]
        print(f"{prod:<25} | {unids:<10} | S/ {ing:<13.2f}")


# Ejecución
ruta_archivo = r"C:\Users\LENOVO\Downloads\ventas_polos_aqp.txt"

with open(ruta_archivo, "r", encoding="utf-8") as archivo:
    lineas_input = archivo.readlines()
    procesar_ventas(lineas_input)
```
